```python
driver_ids = [f"D{i:03d}" for i in range(1, 181)]
```

```python
experience_ranges = [
    (0,2),
    (3,5),
    (6,10),
    (11,20)
]

experience_probabilities = [0.30,0.35,0.25,0.10]
```

```python
license_types = [
    "LMV",
    "Transport",
    "Heavy"
]
license_probabilities = [0.35, 0.45, 0.20]
```

```python
license_data = np.random.choice(
    license_types,
    180,
    p=license_probabilities
)
```

```python
pd.Series(license_data).value_counts()
```

```python
employment_types = [
    "Full-time",
    "Contract",
    "Part-time"
]
employment_probabilities = [0.70, 0.20, 0.10]
```

```python
employment_data=np.random.choice(employment_types,180,p=employment_probabilities)
```

```python
salary_ranges = {
    (0, 2): (22000, 28000),
    (3, 5): (28000, 36000),
    (6, 10): (36000, 48000),
    (11, 20): (48000, 65000)
}
```

```python
driver_experience = []

for _ in range(180):
    experience_range = np.random.choice(
        len(experience_ranges),
        p=experience_probabilities
    )
    
    low, high = experience_ranges[experience_range]
    experience = np.random.randint(low, high + 1)
    driver_experience.append(experience)
```

```python
len(driver_experience)
```

```python
salaries = []

for experience in driver_experience:
    if experience <= 2:
        low, high = salary_ranges[(0, 2)]
    elif experience <= 5:
        low, high = salary_ranges[(3, 5)]
    elif experience <= 10:
        low, high = salary_ranges[(6, 10)]
    else:
        low, high = salary_ranges[(11, 20)]

    salary = np.random.randint(low, high + 1)
    salaries.append(salary)
```

```python
len(salaries)
```

```python
salaries[:10]
```

```python
route_ids = [f"R{i:03d}" for i in range(1, 101)]
```

```python
locations = [
    "Central",
    "North",
    "South",
    "East",
    "West"
]
```

```python
destinations = np.random.choice(locations, 100)

for i in range(100):
    while destinations[i] == origins[i]:
        destinations[i] = np.random.choice(locations)
```

```python
(origins == destinations).sum()
```

```python
distances = np.random.randint(3, 101, 100)
```

```python
road_types = [
    "Urban",
    "Highway",
    "Mixed"
]

road_probabilities = [0.50, 0.25, 0.25]
```

```python
road_type_data = np.random.choice(
    road_types,
    100,
    p=road_probabilities
)
```

```python
road_conditions = [
    "Good",
    "Average",
    "Poor"
]

road_condition_probabilities = [0.60, 0.30, 0.10]
```

```python
road_condition_data = np.random.choice(
    road_conditions,
    100,
    p=road_condition_probabilities
)
```

```python
expected_travel_time = (distances / 40) * 60
expected_travel_time = expected_travel_time * np.random.uniform(
    0.9,
    1.2,
    100
)
```

```python
routes = pd.DataFrame({
    "route_id": route_ids,
    "origin": origins,
    "destination": destinations,
    "distance_km": distances,
    "road_type": road_type_data,
    "road_condition": road_condition_data,
    "expected_travel_time_min": expected_travel_time.round(1)
})
```

```python
routes.head()
```

```python
routes.info()
```

```python
routes.isnull().sum()
```

```python
routes.to_csv("../data/raw/routes.csv", index=False)
```
