```python
package_types=["Small" , "Medium" , "Large" , "Fragile"]
```

```python
package_data=np.random.choice(package_types,num_orders,p=package_probabilities)
```

```python
package_weight_ranges = {
    "Small": (1, 10),
    "Medium": (11, 30),
    "Large": (31, 100),
    "Fragile": (1, 20)
}
```

```python
package_weights = []
for package_type in package_data:
    low, high = package_weight_ranges[package_type]
    weight = np.random.randint(low, high + 1)
    package_weights.append(weight)
    
    
```

```python
# --- Vectorized Tiered Vehicle Assignment ---
weights = np.array(package_weights)

# 1. Map weight brackets to appropriate vehicle capacity
# Small (<=15kg) -> Electric Van (300kg)
# Medium-low (16-30kg) -> Small Van (500kg)
# Medium-high (31-55kg) -> Cargo Car (800kg)
# Heavy (56-80kg) -> Medium Truck (1200kg)
# Very Heavy (>80kg) -> Large Truck (2000kg)
conditions = [
    weights <= 15,
    (weights > 15) & (weights <= 30),
    (weights > 30) & (weights <= 55),
    (weights > 55) & (weights <= 80),
    weights > 80
]
target_capacities = [300, 500, 800, 1200, 2000]

# 2. Extract vehicle pools for each capacity (done once)
vehicle_pools = {
    cap: vehicles[vehicles["max_load_kg"] == cap]["vehicle_id"].to_numpy()
    for cap in target_capacities
}

# 3. Sample vehicles in bulk per tier (takes < 0.01s for 100,000 records)
order_vehicle_ids = np.empty(num_orders, dtype=object)

for cond, cap in zip(conditions, target_capacities):
    count = cond.sum()
    if count > 0:
        order_vehicle_ids[cond] = np.random.choice(vehicle_pools[cap], size=count)

order_vehicle_ids = order_vehicle_ids.tolist()
```

```python
vehicle_license_map = {
    "Electric Van": "LMV",
    "Small Van": "LMV",
    "Cargo Car": "Transport",
    "Medium Truck": "Heavy",
    "Large Truck": "Heavy"
}
```

```python
vehicle_type_map = vehicles.set_index("vehicle_id")["vehicle_type"]
```

```python
# --- Vectorized Driver Assignment ---
# 1. Pre-group drivers by license once
driver_pools = drivers.groupby("license_type")["driver_id"].apply(np.array).to_dict()

# 2. Fast mapping from vehicle_id -> required license
vehicle_to_license = vehicles.set_index("vehicle_id")["vehicle_type"].map(vehicle_license_map).to_dict()
order_licenses = np.array([vehicle_to_license[vid] for vid in order_vehicle_ids])

# 3. Vectorized sampling per license type
order_driver_ids = np.empty(num_orders, dtype=object)

for lic_type, pool in driver_pools.items():
    mask = (order_licenses == lic_type)
    count = mask.sum()
    if count > 0:
        order_driver_ids[mask] = np.random.choice(pool, size=count)

order_driver_ids = order_driver_ids.tolist()
```

```python
time_slots = [
    "Morning",
    "Late Morning",
    "Afternoon",
    "Evening",
    "Night"
]
time_probabilities = [0.20, 0.25, 0.30, 0.20, 0.05]
```
