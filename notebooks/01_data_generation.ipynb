```python
package_types=["Small" , "Medium" , "Large" , "Fragile"]
```

```python
package_probabilities=[0.4,0.35,0.15,0.1]
```

```python
package_data=np.random.choice(package_types,num_orders,p=package_probabilities)
```

```python
package_weight_ranges = {
    "Small": (1, 10),
    "Medium": (11, 30),
    "Large": (31, 100),
    "Fragile": (1, 20)
}
```

```python
package_weights = []
for package_type in package_data:
    low, high = package_weight_ranges[package_type]
    weight = np.random.randint(low, high + 1)
    package_weights.append(weight)
    
    
```

```python
order_customer_ids = np.random.choice(
    customers["customer_id"],
    num_orders
)
```

```python
pd.Series(order_customer_ids).value_counts().head()
```

```python
order_route_ids=np.random.choice(routes["route_id"],num_orders)
```

```python
vehicle_groups = {}

for capacity in vehicles["max_load_kg"].unique():
    vehicle_groups[capacity] = vehicles[
        vehicles["max_load_kg"] == capacity
    ]["vehicle_id"].tolist()
```

```python
order_vehicle_ids = []

capacities = sorted(vehicle_groups.keys())

for weight in package_weights:
    for capacity in capacities:
        if capacity >= weight:
            vehicle_id = np.random.choice(vehicle_groups[capacity])
            order_vehicle_ids.append(vehicle_id)
            break
```

```python
vehicle_license_map = {
    "Electric Van": "LMV",
    "Small Van": "LMV",
    "Cargo Car": "Transport",
    "Medium Truck": "Heavy",
    "Large Truck": "Heavy"
}
```

```python
vehicle_type_map = vehicles.set_index("vehicle_id")["vehicle_type"]
```

```python
order_driver_ids = []

for vehicle_id in order_vehicle_ids:
    vehicle_type = vehicle_type_map[vehicle_id]
    required_license = vehicle_license_map[vehicle_type]

    eligible_drivers = drivers[
        drivers["license_type"] == required_license
    ]

    eligible_ids = eligible_drivers["driver_id"]
    driver_id = np.random.choice(eligible_ids)

    order_driver_ids.append(driver_id)
```

```python
time_slots = [
    "Morning",
    "Late Morning",
    "Afternoon",
    "Evening",
    "Night"
]
time_probabilities = [0.20, 0.25, 0.30, 0.20, 0.05]
```
